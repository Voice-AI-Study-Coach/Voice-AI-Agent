```python
#LayoutParser is the step where we will pass bounding boxes to the model and get the text from it.
import fitz
import os
import layoutparser as lp
import cv2
```

```python
doc = fitz.open("C:\\Users\\HP\\Downloads\\Voice-AI-Agent\\ocr-testing\\DAA UNIT 4 NOTES.pdf")
output_dir = "images"
os.makedirs(output_dir, exist_ok=True)
```

```python

# Increase the DPI to 300 for better quality images
for page_num in range(len(doc)):
    page = doc[page_num]
    pix = page.get_pixmap(dpi=300)
    output_path = os.path.join(output_dir, f"page_{page_num + 1}.png") 
    pix.save(output_path)
doc.close()
```

```python
import layoutparser as lp

# Load the pre-trained model PaddleDetection using PaddleDetectionLayoutModel
# model = lp.PaddleDetectionLayoutModel(
#     config_path='lp://PubLayNet/ppyolov2_r50vd_dcn_365e/config',
#     label_map={0: "Text", 1: "Title", 2: "List", 3: "Table", 4: "Figure"},
#     enforce_cpu=True,
#     extra_config={"threshold": 0.5}  # <-- Pass threshold parameters here
# )
model = lp.PaddleDetectionLayoutModel(
    config_path='lp://PubLayNet/ppyolov2_r50vd_dcn_365e/config',
    extra_config={"threshold": 0.5},
    label_map={0: "Text", 1: "Title", 2: "List", 3: "Table", 4: "Figure"},
    enforce_cpu=True  # set False if you have a working CUDA GPU setup
)
```
