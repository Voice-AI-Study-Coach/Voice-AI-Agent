```python
import os
from openai import OpenAI
```

```python
client = OpenAI(
    base_url="https://router.huggingface.co/v1",
    api_key=os.environ["HUGGING_FACE_TOKEN"],
)
```

```python
completion = client.chat.completions.create(
    model="moonshotai/Kimi-K3:together",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "You are a very intelligent model which answers user's questions"},
                {"type": "text", "text": "Do you know about Fable 5 model?"}
            ]
        }
    ],
)
```

```python
completion
```
